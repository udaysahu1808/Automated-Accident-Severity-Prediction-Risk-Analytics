# Automated Accident Severity Prediction & Risk Analytics

### Project Introduction

Road Traffic Accidents (RTAs) are one of the leading causes of fatalities, injuries, and economic losses worldwide. The increasing volume of accident data generated by transportation agencies provides an opportunity to analyze accident patterns and predict accident severity using data-driven approaches. However, manual Exploratory Data Analysis (EDA) is often time-consuming and may overlook important insights hidden within large datasets.

This project focuses on developing an Automated Accident Severity Prediction and Risk Analytics framework using modern Automated EDA tools such as YData Profiling, AutoViz, and DTale. These tools automate the process of data profiling, visualization, missing value analysis, correlation detection, and interactive data exploration, thereby significantly reducing the time required for data understanding and preprocessing.

The project begins with comprehensive data profiling using YData Profiling to generate detailed HTML reports containing descriptive statistics, missing value summaries, correlations, and warnings. AutoViz is employed to automatically generate a wide variety of visualizations, while DTale provides an interactive graphical interface for filtering, sorting, and analyzing data in real time.

After automated data exploration, machine learning techniques are applied to predict accident severity levels and identify key risk factors contributing to severe accidents. The project also compares the strengths and limitations of different automated EDA tools and evaluates their effectiveness in accelerating the data analysis workflow.

This study demonstrates how automated data exploration combined with predictive analytics can improve accident risk assessment, support decision-making, and enhance road safety initiatives.

### Phase 7: Business Insights

##### 1. Driver-Related Factors Significantly Influence Accident Severity
The analysis reveals that driver characteristics such as age, driving experience, and driving behavior are among the most influential factors affecting accident severity. Younger and less experienced drivers tend to be involved in more severe accidents, indicating the need for focused safety training and awareness programs.

##### 2. Certain Vehicle Types are Associated with Higher Accident Risks
Heavy vehicles and commercial vehicles contribute to a larger proportion of severe accidents compared to smaller passenger vehicles. This suggests that stricter safety regulations and periodic inspections are essential for high-risk vehicle categories.

##### 3. Adverse Weather Conditions Increase Accident Severity
Weather conditions such as rain, fog, and poor visibility have a significant impact on accident severity. Accidents occurring during unfavorable weather conditions are generally more severe due to reduced visibility and slippery road surfaces.

##### 4. Poor Road Surface Conditions Contribute to Severe Accidents
Roads with damaged surfaces, potholes, or slippery conditions show a higher frequency of serious accidents. Maintaining road quality and ensuring proper drainage systems can help reduce accident severity.

##### 5. Night-Time Driving Carries Higher Risk
The analysis indicates that accidents occurring during dark or low-light conditions are generally more severe. Insufficient street lighting and reduced visibility increase the probability of major injuries and fatalities.

##### 6. Human Errors Remain the Leading Cause of Accidents
The majority of accidents are caused by driver-related factors such as speeding, reckless driving, distraction, and violation of traffic rules. Behavioral interventions can therefore play a critical role in accident prevention.

##### 7. Automated EDA Tools Save Significant Analysis Time
YData Profiling, AutoViz, and D-Tale automate exploratory data analysis, reducing the time required for data understanding, visualization, and quality assessment while improving analytical efficiency.

##### 8. High-Risk Accident Scenarios Can Be Identified Effectively
The risk analytics framework helps identify combinations of factors such as poor weather, inexperienced drivers, and adverse road conditions that contribute to severe accidents, enabling proactive risk management.

##### 9. Data Quality Assessment Improves Analytical Reliability
Automated profiling identifies missing values, duplicate records, and outliers efficiently, ensuring more reliable and accurate accident analysis.

##### 10. Data-Driven Road Safety Strategies are More Effective
Insights generated through automated analytics can support government agencies and transportation authorities in designing evidence-based road safety policies and accident prevention strategies.
